Dataset 1: Penguins – EDA

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style="whitegrid")
```

```python
url = "https://raw.githubusercontent.com/rerkki/data-analytics-ml/refs/heads/main/datasets/penguins_size.csv"
```

```python
url = "https://raw.githubusercontent.com/rerkki/data-analytics-ml/refs/heads/main/datasets/penguins_size.csv"
df = pd.read_csv(url)
df.head()
```

```python
df.shape
df.info()
df.isna().sum()
```

```python
df.describe()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["body_mass_g"], kde=True)
plt.title("Distribution of Body Mass")
plt.show()

plt.figure(figsize=(8,5))
sns.boxplot(x="species", y="body_mass_g", data=df)
plt.title("Body Mass by Species")
plt.show()
```

```python
df.columns
```

```python
numeric_cols = ["culmen_length_mm", "culmen_depth_mm", "flipper_length_mm", "body_mass_g"]
corr = df[numeric_cols].corr()

plt.figure(figsize=(6,4))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

In the Penguins dataset, there are three species with clear differences in body mass and flipper length. The distribution of body mass is roughly normal with some outliers. There is a strong positive correlation between flipper length and body mass, meaning larger penguins tend to be heavier.

Dataset 2: Titanic – EDA

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df_titanic = pd.read_csv(url)
df_titanic.head()
```

```python
df_titanic.shape
df_titanic.info()
df_titanic.isna().sum()
```

```python
df_titanic.describe()
```

```python
sns.histplot(df_titanic["Age"], kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
sns.boxplot(x="Pclass", y="Fare", data=df_titanic)
plt.title("Fare by Passenger Class")
plt.show()
```

```python
numeric_cols = ["Age", "Fare", "SibSp", "Parch"]
corr = df_titanic[numeric_cols].corr()

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

The Titanic dataset shows clear differences in fare between passenger classes. Age distribution is right-skewed with missing values. Fare has strong positive correlation with passenger class. Some outliers exist in fare values

```python
sns.boxplot(x="species", y="sepal_length", data=df_iris)
plt.title("Sepal Length by Species")
plt.show()
```

The Iris dataset shows clear differences between species in both sepal and petal measurements. Petal length and petal width have a strong positive correlation. Sepal measurements vary less across species. The dataset is clean, with no missing values, and suitable for classification tasks

Dataset 5: Wine Quality – EDA

```python
url = "https://raw.githubusercontent.com/plotly/datasets/master/winequality-red.csv"
df_wine = pd.read_csv(url)
df_wine.head()
```

```python
df_wine.shape
```

```python
df_wine.info()
```

```python
df_wine.isna().sum()
```

```python
df_wine.describe()
```

```python
sns.histplot(df_wine["alcohol"], kde=True)
plt.title("Distribution of Alcohol")
plt.show()
```

```python
sns.histplot(df_wine["quality"], kde=False)
plt.title("Distribution of Wine Quality")
plt.show()
```

```python
sns.boxplot(y=df_wine["alcohol"])
plt.title("Alcohol Outliers")
plt.show()
```

```python
sns.boxplot(x="quality", y="alcohol", data=df_wine)
plt.title("Alcohol by Quality")
plt.show()
```

```python
numeric_cols = df_wine.select_dtypes(include=['float64', 'int64']).columns
corr = df_wine[numeric_cols].corr()

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

```python
sns.boxplot(y=df_wine["residual sugar"])
plt.title("Outliers in Residual Sugar")
plt.show()
```

The Wine Quality dataset shows that alcohol level has a positive correlation with wine quality. Volatile acidity is negatively correlated with quality. Several chemical features show skewed distributions and contain outliers. The dataset is suitable for regression and classification tasks
